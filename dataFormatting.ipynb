```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("rawData.csv")
df['Filename'] = df['Filename'].str.extract(r'(\d+)').astype(int)
df = df.sort_values(by='Filename').reset_index(drop=True)
df.head()
```

```python
subject_name = df["Folder"].unique()
subject_name
```

```python
for name in subject_name:
    match name:
        case 'Vu': 
            df.loc[df['Folder'] == name,'Take'] = 2
            df.loc[df['Folder'] == name,'Not take'] = 1

        case 'Minh':
            df.loc[df['Folder'] == name,'Take'] = 6
            df.loc[df['Folder'] == name,'Not take'] = 3

        case 'tuda_real':
            df.loc[df['Folder'] == name,'Take'] = 3
            df.loc[df['Folder'] == name,'Not take'] = 5

        case 'Phuc':
            df.loc[df['Folder'] == name,'Take'] = 2
            df.loc[df['Folder'] == name,'Not take'] = 1

df.head(10)
```

```python
df.to_csv("formattedData.csv", index = False)
```
