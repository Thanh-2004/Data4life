```python
import pandas as pd
import numpy as np
import os 
import json
```

```python
df = pd.read_csv("formattedData.csv")
df.head()
```

```python
subject_name = df["Folder"].unique()
subject_name
```

```python
parent_folder = "Data"
folder_paths = [os.path.join(parent_folder, folder) for folder in os.listdir(parent_folder) if os.path.isdir(os.path.join(parent_folder, folder))]
folder_paths.remove('Data/duc')

print(folder_paths)

```

```python
results = []

for folder in folder_paths:
    for filename in os.listdir(folder):
        if filename.endswith(".json"):
            file_path = os.path.join(folder, filename)
            print(file_path)
            # Đọc dữ liệu từ file txt và lưu vào list
            with open(file_path, 'r') as file:
                result = json.load(file)
                results.append(result)
```

```python
all_images = []
all_actions = []

for result in results:
    all_images.append([item['image'] for item in result])
    all_actions.append([item['action'] for item in result])

for image in all_images:
    print(len(image))

```

```python
for name in subject_name:
    match name:
        case 'Vu': 
            df.loc[df['Folder'] == name,'Image'] = all_images[0]
            df.loc[df['Folder'] == name,'Action'] = all_actions[0]

        case 'Minh':
            df.loc[df['Folder'] == name,'Image'] = all_images[1]
            df.loc[df['Folder'] == name,'Action'] = all_actions[1]

        case 'tuda_real':
            df.loc[df['Folder'] == name,'Image'] = all_images[2]
            df.loc[df['Folder'] == name,'Action'] = all_actions[2]

        case 'Phuc':
            df.loc[df['Folder'] == name,'Image'] = all_images[3]
            df.loc[df['Folder'] == name,'Action'] = all_actions[3]
```

```python
df.loc[df["Folder"] == "Phuc"]
```

```python
df['Image'] = df['Image'].str.extract(r'(\d+)').astype(float)
df.head()
```

```python
df['P300'] = df.apply(lambda row: 'Yes' if row['Image'] == row['Take'] or row['Image'] == row['Not take'] else 'No', axis=1)
df.head()
```

```python
mapping = {"Yes": 1, "No": 0}
df['Action'] = df['Action'].map(mapping)
df['P300'] = df['P300'].map(mapping)

df.head()
```

```python
df['Deception'] = df['Action'] ^ df['P300']
df.head()
```

```python
deception_mapping = {0: "Truth", 1: "Lie"}
df['Deception'] = df['Deception'].map(deception_mapping)
df.head()
```

```python
df.to_csv("finalData.csv", index = False)
```
